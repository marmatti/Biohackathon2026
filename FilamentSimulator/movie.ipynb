```python
from pathlib import Path
import json
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
import matplotlib.image as mpimg
from IPython.display import HTML
```

```python
out_dir = Path("outputs/base/run_000_seed_164855472")
clean_dir = out_dir / "clean_tiff_frames"
noisy_dir = out_dir / "noisy_tiff_frames"

print("Clean dir:", clean_dir.resolve())
print("Noisy dir:", noisy_dir.resolve())
print("Clean exists:", clean_dir.exists())
print("Noisy exists:", noisy_dir.exists())
```

```python
clean_files = sorted(clean_dir.glob("clean_*.tif"))
noisy_files = sorted(noisy_dir.glob("noisy_*.tif"))
with open(out_dir / "frame_labels.json", "r") as f:
    frame_labels = json.load(f)


print("Number of clean frames:", len(clean_files))
print("Number of noisy frames:", len(noisy_files))

assert len(clean_files) > 0, "No clean TIF files found."
assert len(noisy_files) > 0, "No noisy TIF files found."
assert len(clean_files) == len(noisy_files), "Clean/noisy frame counts do not match."

clean_imgs = [mpimg.imread(p) for p in clean_files]
noisy_imgs = [mpimg.imread(p) for p in noisy_files]
thick_masks = [
    np.array(frame_labels[i]["filament_mask_2d"], dtype=np.uint8)
    for i in range(len(frame_labels))
]

thick_masks = [
    np.array(frame_labels[i]["filament_mask_2d"], dtype=np.uint8)
    for i in range(len(frame_labels))
]

print("Clean frame shape:", clean_imgs[0].shape)
print("Noisy frame shape:", noisy_imgs[0].shape)
```

```python
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML

fig, axes = plt.subplots(1, 3, figsize=(15, 5))

im_clean = axes[0].imshow(clean_imgs[0], cmap="gray")
axes[0].set_title("Clean")
axes[0].axis("off")

im_mask = axes[1].imshow(thick_masks[0], cmap="Reds", vmin=0, vmax=1)
axes[1].set_title("Filament Mask")
axes[1].axis("off")

im_noisy = axes[2].imshow(noisy_imgs[0], cmap="gray")
axes[2].set_title("Noisy")
axes[2].axis("off")

suptitle = fig.suptitle("Frame 0")

def update(i):
    im_clean.set_data(clean_imgs[i])
    im_mask.set_data(thick_masks[i])
    im_noisy.set_data(noisy_imgs[i])
    suptitle.set_text(f"Frame {i}")
    return im_clean, im_mask, im_noisy

ani = FuncAnimation(
    fig,
    update,
    frames=len(clean_imgs),
    interval=100,
    blit=False,
    repeat=True,
)

plt.close(fig)
HTML(ani.to_jshtml())
```
